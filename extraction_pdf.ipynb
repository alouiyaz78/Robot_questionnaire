```python
import pdfplumber
import os

# Chemin vers ton PDF
pdf_path = "Cours 3 Fusion et préparation des données.pdf"

# Vérifier que le fichier existe
if not os.path.exists(pdf_path):
    print("Le fichier PDF n'existe pas.")
    exit()

# Créer le dossier Data si nécessaire
data_dir = "Data"
os.makedirs(data_dir, exist_ok=True)

# Créer un nom de fichier texte de base
base_name = os.path.splitext(os.path.basename(pdf_path))[0]
txt_path = os.path.join(data_dir, base_name + ".txt")

# Si le fichier existe déjà, ajouter un suffixe
counter = 1
while os.path.exists(txt_path):
    txt_path = os.path.join(data_dir, f"{base_name}_{counter}.txt")
    counter += 1

# Extraire le texte
text = ""
with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
        page_text = page.extract_text()
        if page_text:
            text += page_text + "\n"

# Enregistrer le texte
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(text)

print(f"Texte extrait et enregistré dans : {txt_path}")
```
